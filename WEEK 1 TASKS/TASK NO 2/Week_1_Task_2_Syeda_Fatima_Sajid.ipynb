**Program 1: Add Two Numbers**

```python
# Program to add two numbers

num1 = 10
num2 = 5

sum_result = num1 + num2

print("First number:", num1)
print("Second number:", num2)
print("Sum is:", sum_result)

#Modify Values
num1 = 55
num2 = 1

sum_result = num1 + num2

print("Result After Modifying Values:")
print("First number:", num1)
print("Second number:", num2)
print("Sum is:", sum_result)

```

Explanation

This program adds two numbers using the + operator.
When num1 = 10 and num2 = 5, the output is 15.
After modifying the values, the output changes accordingly.

**Program 2: Calculate Average of 5 Numbers**

```python
# Program to calculate average of 5 numbers

n1 = 10
n2 = 20
n3 = 30
n4 = 40
n5 = 50

average = (n1 + n2 + n3 + n4 + n5) / 5

print("Numbers:", n1, n2, n3, n4, n5)
print("Average is:", average)
#Modify Values
n1 = 1
n2 = 2
n3 = 3
n4 = 4
n5 = 5

average = (n1 + n2 + n3 + n4 + n5) / 5

print("Numbers:", n1, n2, n3, n4, n5)
print("Average is:", average)
```

**Explanation**
This program calculates the average of five numbers by adding them and dividing by 5.
When the numbers are 10, 20, 30, 40, and 50, the average is 30.
After modifying the numbers, the average changes accordingly.

**Program 3: Check Even or Odd**

```python
# Program to check even or odd number

number = 8

if number % 2 == 0:
    print(number, "is Even")
else:
    print(number, "is Odd")
#Modify Values

number = 11

if number % 2 == 0:
    print(number, "is Even")
else:
    print(number, "is Odd")
```

**Explanation**

This program checks whether a number is even or odd using the modulus operator (%).
If the remainder when dividing by 2 is 0, the number is even.
Otherwise, it is odd.
When number = 8, the output shows Even.
When changed to 11, it shows Odd.
